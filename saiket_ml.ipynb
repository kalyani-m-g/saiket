```python
df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0})
```

```python
df.drop("customerID", axis=1, inplace=True)
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
df.head()
```

```python
df.to_csv("Preprocessed_Churn_Data.csv", index=False)
```

```python
##task1 report:The Telco Customer Churn dataset was loaded into Google Colab and examined using Pandas functions. Missing values were checked and none were found in the dataset. The target variable `Churn` was converted into numerical format (Yes = 1, No = 0). The `customerID` column was removed as it is not useful for prediction. Categorical variables were encoded into numerical values using one-hot encoding. The dataset was successfully preprocessed and made ready for machine learning tasks.
```

```python
#Task 2: Split Data for Training and Testing.
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop("Churn", axis=1)
y = df["Churn"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
```

```python
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

```python
#task 2 report: The dataset was successfully divided into training and testing sets using an 80:20 ratio. Stratified sampling was applied to maintain the original distribution of the target variable (Churn) in both datasets. The training set contained approximately 73.46% non-churn customers and 26.54% churn customers, while the testing set showed a nearly identical distribution. This ensures representative data for model training and evaluation.
```

```python
#TASK 3: FEATURE SELECTION
```

```python
correlation = df.corr()["Churn"].sort_values(ascending=False)
print(correlation)
```

```python
print(correlation.head(15))
```

```python
selected_features = [
    'tenure',
    'MonthlyCharges',
    'Contract_One year',
    'Contract_Two year',
    'OnlineSecurity_Yes',
    'TechSupport_Yes',
    'InternetService_Fiber optic',
    'PaperlessBilling_Yes'
]
```

```python
X_selected = df[selected_features]
X_selected.head()
```

```python
print(X_selected.columns)
```

```python
#task 3 report: Feature selection was performed to identify the attributes that significantly influence customer churn. Correlation analysis was used to examine the relationship between features and the target variable (Churn). Based on the analysis, important features such as tenure, monthly charges, contract type, online security, tech support, internet service type, and paperless billing were selected. These features were chosen for model training as they have a strong impact on customer churn prediction.
```

```python
#TASK 4: MODEL SELECTION
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
```

```python
print(model)
```

```python
#task 4 report: A suitable binary classification algorithm was selected for customer churn prediction. Logistic Regression was chosen because the target variable (Churn) has two classes: churn and non-churn. The algorithm is simple, efficient, and widely used for binary classification problems. It provides good performance and helps understand the relationship between customer attributes and churn behavior.
```

```python
#TASK 5: MODEL TRAINING
```

```python
model.fit(X_train[selected_features], y_train)
```

```python
print("Model trained successfully!")
```

```python
print(model.coef_)
```

```python
y_pred = model.predict(X_test[selected_features])
```

```python
print(y_pred[:10])
```

```python
#task 5 report : The Logistic Regression model was trained using the training dataset. The selected features were used as input variables, and the Churn column was used as the target variable. The model learned patterns from historical customer data and was successfully fitted to the training set. After training, the model was used to generate predictions for customer churn on the testing dataset.
```

```python
#TASK 6: MODEL EVALUATION
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred)
```

```python
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
print("ROC-AUC:", roc_auc)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
#task 6 report : The performance of the Logistic Regression model was evaluated using the testing dataset. The model achieved an accuracy of 78%. For churn prediction, the precision was 60%, recall was 50%, and F1-score was 55%. The model performed well in identifying non-churn customers, with an F1-score of 85% for class 0. Overall, the model demonstrated satisfactory performance and can be used to predict customer churn with reasonable accuracy.
```

```python
sample_customer = X_test[selected_features].iloc[[0]]

print("Actual:", y_test.iloc[0])
print("Predicted:", model.predict(sample_customer)[0])
print("Churn Probability:", model.predict_proba(sample_customer)[0][1])
```

```python
#inference: A sample customer from the test dataset was given as input to the trained Logistic Regression model. The model predicted class 0 (No Churn) with a churn probability of 4.4%. The actual value was also 0, indicating that the prediction was correct. This demonstrates that the model can successfully classify customers based on their attributes and estimate the likelihood of churn.
```

```python
#churn example
for i in range(len(X_test)):
    sample = X_test[selected_features].iloc[[i]]

    pred = model.predict(sample)[0]

    if pred == 1:
        print("Index:", i)
        print("Actual:", y_test.iloc[i])
        print("Predicted:", pred)
        print("Churn Probability:", model.predict_proba(sample)[0][1])
        break
```
